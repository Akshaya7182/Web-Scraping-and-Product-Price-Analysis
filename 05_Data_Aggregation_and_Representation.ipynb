## Stage Objective

Stage 5 performs grouping, counting and aggregation so that large numbers of individual product records can be summarized meaningfully.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path = "/content/Air_Conditioners_Cleaned.csv"
df = pd.read_csv(csv_path)

print("CSV loaded successfully.")
print("Dataset shape:", df.shape)
```

## Why Aggregation Is Needed

Individual rows answer product-level questions. Aggregation answers questions about groups, such as how many products belong to a category or what the average numeric value is for a group.


```python
cats=df.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
```

## Reload the CSV

The dataset is loaded independently and the fields required for aggregation are identified.


```python
nums=df.select_dtypes(include=np.number).columns.tolist(); print(nums)
```

## Choose Grouping Fields

Categorical fields are examined and an actual field from the dataset is selected for group-level summaries.


```python
if cats: counts=df.groupby(cats[0],dropna=False).size().reset_index(name='count'); print(counts)
```

## Count by Category

The number of records in each available category is calculated.


```python
if cats and nums: summary=df.groupby(cats[0],dropna=False)[nums[0]].agg(['count','mean','min','max','sum']).reset_index(); print(summary)
```

## Calculate Numeric Summaries

For a suitable numeric field, count, mean, minimum, maximum and sum are calculated by group.


```python
if 'summary' in locals(): print(summary.sort_values('mean',ascending=False))
```

## Sort the Summary

The aggregated table is sorted so differences between groups are easier to inspect.


```python
if cats: print(df[cats[0]].value_counts(dropna=False).head(15))
```

## Compare Group Sizes

Category counts show where the dataset has more or fewer observations. This helps contextualize later averages.


```python
if 'summary' in locals(): print(summary.describe(include='all').T)
```

## Create Reusable Summary Tables

Compact summary DataFrames are created so Stage 6 can analyse them and Stage 7 can visualise them.


```python
if 'summary' in locals(): print('Summary rows:',len(summary))
```

The summary table is inspected for shape and values so grouping errors are easier to detect.
