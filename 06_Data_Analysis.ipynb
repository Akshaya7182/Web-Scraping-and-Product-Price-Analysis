## Stage Objective

Stage 6 performs descriptive analysis on the validated product data and the summary tables produced earlier.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path = "/content/Air_Conditioners_Cleaned.csv"
df = pd.read_csv(csv_path)

print("CSV loaded successfully.")
print("Dataset shape:", df.shape)
```

## What Analysis Means Here

The project uses descriptive analysis: we measure and compare what is present in the supplied dataset. We do not claim that an observed association proves causation.


```python
nums=df.select_dtypes(include=np.number).columns.tolist(); print(nums)
```

## Reload the CSV

The dataset is loaded directly so the analysis notebook is independently reproducible.


```python
print(df[nums].describe().T if nums else 'No numeric columns found.')
```

## Overall Statistics

Descriptive statistics are generated for numerical fields to understand typical values and the observed spread.


```python
print(df['discount_price'].describe() if 'discount_price' in df.columns else 'Price field not detected.')
```

## Price Analysis

The available price field is **discount_price**. Its distribution, range and summary statistics are examined.


```python
print(df['ratings'].describe() if 'ratings' in df.columns else 'Rating field not detected.')
```

## Rating Analysis

The available rating field is **ratings**. Its range and distribution are examined.


```python
cats=df.select_dtypes(include=['object','category','bool']).columns.tolist()
for c in cats[:5]: print(c,df[c].value_counts().head(10).to_dict())
```

## Category Comparisons

Group-level results are compared to understand how product attributes differ between available categories.


```python
for c in nums[:5]: print('Top',c); print(df[[c]].sort_values(c,ascending=False).head(5))
```

## High and Low Observations

Sorted results are used to inspect high and low values and connect the numerical summaries to concrete product records.


```python
print(df[nums].corr() if len(nums)>=2 else 'Need at least two numeric fields for correlation.')
```
