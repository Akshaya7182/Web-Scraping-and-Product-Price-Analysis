**Stage 2: Data Acquisition & Filtering**

This notebook is independently executable in Google Colab. Every step explains what we do, why we do it, what we obtain, and how the result is used in the next stage.


## Stage Objective

Stage 2 focuses on obtaining useful subsets of the dataset through transparent filtering and sorting operations.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path1 = "/content/Most_Expensive_Products.csv"
df1 = pd.read_csv(csv_path1)
csv_path2 = "/content/Top_Discounted_Products.csv"
df2 = pd.read_csv(csv_path2)
csv_path3 = "/content/Top_Rated_Products.csv"
df3 = pd.read_csv(csv_path3)

print("CSV loaded successfully.")
print("Dataset shape1:", df1.shape)
print("Dataset shape2:", df2.shape)
print("Dataset shape3:", df3.shape)
```

## Why Filtering Is Needed

The full product table is useful for overview, but individual analytical questions often require a focused subset. Filtering lets us isolate records without altering the original source.


```python
cats=df1.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
cats=df2.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
cats=df3.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
```

## Reload the CSV

The dataset is loaded again with `read_csv()` so this stage can be executed independently in Google Colab.


```python
nums=df1.select_dtypes(include=np.number).columns.tolist(); print(nums)
nums=df2.select_dtypes(include=np.number).columns.tolist(); print(nums)
nums=df3.select_dtypes(include=np.number).columns.tolist(); print(nums)
```

## Inspect Categories

Available categorical fields and their values are inspected before filtering. This ensures that filters use categories that actually exist in the supplied data.


```python
for c in nums[:5]: print(c, df1[c].min(), df1[c].max())
for c in nums[:5]: print(c, df2[c].min(), df2[c].max())
for c in nums[:5]: print(c, df3[c].min(), df3[c].max())
```

## Inspect Numeric Ranges

Numeric fields are examined to understand their observed ranges before applying any numerical filter.


```python
if cats: print(df1[cats[0]].value_counts(dropna=False).head(15))
if cats: print(df2[cats[0]].value_counts(dropna=False).head(15))
if cats: print(df3[cats[0]].value_counts(dropna=False).head(15))
```

## Category-Based View

A category-based subset is created using an actual categorical column from the dataset. The resulting row count shows how many records belong to that view.


```python
if cats: filtered=df1[df1[cats[0]].notna()].copy(); print(filtered.shape)
if cats: filtered=df2[df2[cats[0]].notna()].copy(); print(filtered.shape)
if cats: filtered=df3[df3[cats[0]].notna()].copy(); print(filtered.shape)
```

## Numeric-Based View

Where numeric attributes are available, a range-based subset is demonstrated using observed values rather than arbitrary assumptions.


```python
if nums: print(df1.sort_values(nums[0],ascending=False).head(10)[nums[0]])
if nums: print(df2.sort_values(nums[0],ascending=False).head(10)[nums[0]])
if nums: print(df3.sort_values(nums[0],ascending=False).head(10)[nums[0]])
```

## Sorting

Records are sorted using a numerical attribute so that high and low observations can be inspected directly.


```python
print('Original:',len(df1),'Filtered:',len(filtered) if 'filtered' in locals() else len(df1))
print('Original:',len(df2),'Filtered:',len(filtered) if 'filtered' in locals() else len(df2))
print('Original:',len(df3),'Filtered:',len(filtered) if 'filtered' in locals() else len(df3))
```

## Compare Row Counts

Original and filtered row counts are compared so the effect of each filtering operation is measurable.


```python
analysis_view=(filtered if 'filtered' in locals() else df1).copy(); print(analysis_view.head())
analysis_view=(filtered if 'filtered' in locals() else df2).copy(); print(analysis_view.head())
analysis_view=(filtered if 'filtered' in locals() else df3).copy(); print(analysis_view.head())
```
