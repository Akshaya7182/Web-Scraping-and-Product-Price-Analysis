## Stage Objective

This stage establishes the starting point of the Air Conditioners analysis. We first load the supplied CSV and verify exactly what data is available before performing any filtering or interpretation.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path1 = "/content/Most_Expensive_Products.csv"
df1 = pd.read_csv(csv_path1)
csv_path2 = "/content/Top_Discounted_Products.csv"
df2 = pd.read_csv(csv_path2)
csv_path3 = "/content/Top_Rated_Products.csv"
df3 = pd.read_csv(csv_path3)

print("CSV loaded successfully.")
print("Dataset shape:", df1.shape)
print("Dataset shape:", df2.shape)
print("Dataset shape:", df3.shape)
```

```python
print('Rows1:', df1.shape[0]); print('Columns1:', df1.shape[1])
print('Rows2:', df2.shape[0]); print('Columns2:', df2.shape[1])
print('Rows3:', df3.shape[0]); print('Columns3:', df3.shape[1])
```

```python
print(df1.columns.tolist())
print(df2.columns.tolist())
print(df3.columns.tolist())
```

```python
print(df1.dtypes)
print(df2.dtypes)
print(df3.dtypes)
```
