## Air Conditioners Data Analysis

**Stage 3: Data Extraction**

This notebook is independently executable in Google Colab. Every step explains what we do, why we do it, what we obtain, and how the result is used in the next stage.


## Stage Objective

Stage 3 extracts the fields required for analysis and creates focused representations of the product data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path1 = "/content/Most_Expensive_Products.csv"
df1 = pd.read_csv(csv_path1)
csv_path2 = "/content/Top_Discounted_Products.csv"
df2 = pd.read_csv(csv_path2)
csv_path3 = "/content/Top_Rated_Products.csv"
df3 = pd.read_csv(csv_path3)

print("CSV loaded successfully.")
print("Dataset shape:", df1.shape)
print("Dataset shape:", df2.shape)
print("Dataset shape:", df3.shape)
```

## Why Extraction Matters

Working with every field at every step makes analysis harder to explain. Selecting relevant fields creates smaller, purpose-specific views.


```python
nums=df1.select_dtypes(include=np.number).columns.tolist(); print(nums)
nums=df2.select_dtypes(include=np.number).columns.tolist(); print(nums)
nums=df3.select_dtypes(include=np.number).columns.tolist(); print(nums)
```

## Reload the CSV

The CSV is read directly so this stage remains self-contained.


```python
cats=df1.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
cats=df2.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
cats=df3.select_dtypes(include=['object','category','bool']).columns.tolist(); print(cats)
```

## Identify Numeric Fields

The project contains numeric fields such as: **ratings, no_of_ratings, discount_price, actual_price, discount_amount, discount_percent, tonnage, energy_star**. These fields can support descriptive statistics and numerical comparisons.


```python
numeric_view=df1[nums].copy(); print(numeric_view.shape)
numeric_view=df2[nums].copy(); print(numeric_view.shape)
numeric_view=df3[nums].copy(); print(numeric_view.shape)
```

## Identify Categorical Fields

The project contains categorical/text fields such as: **name, main_category, sub_category, image, link, brand**. These fields can support grouping and category-level comparisons.


```python
categorical_view=df1[cats].copy(); print(categorical_view.shape)
categorical_view=df2[cats].copy(); print(categorical_view.shape)
categorical_view=df3[cats].copy(); print(categorical_view.shape)
```

## Extract Numeric View

A numeric-only view is created to support later statistical analysis without unnecessary text columns.


```python
print(numeric_view.head())
print(numeric_view.head())
print(numeric_view.head())
```

## Extract Categorical View

A categorical view is created to support frequency counts and group comparisons.


```python
print(categorical_view.head())
print(categorical_view.head())
print(categorical_view.head())
```

## Inspect Extracted Views

Shapes and sample records are displayed so we can confirm that extraction preserved the intended observations.


```python
for c in nums[:5]: print(c, numeric_view[c].describe().to_dict())
for c in nums[:5]: print(c, numeric_view[c].describe().to_dict())
for c in nums[:5]: print(c, numeric_view[c].describe().to_dict())
```

## Derived Analytical Views

Where appropriate, focused combinations of fields can be used for specific questions, such as comparing a product's price with its rating or category.


```python
for c in cats[:5]: print(c, categorical_view[c].nunique(dropna=False))
for c in cats[:5]: print(c, categorical_view[c].nunique(dropna=False))
for c in cats[:5]: print(c, categorical_view[c].nunique(dropna=False))
```

## What We Obtained

The output of this stage is a set of clearly defined analytical views rather than an unstructured collection of columns.
