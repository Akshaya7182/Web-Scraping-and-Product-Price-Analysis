## Air Conditioners Data Analysis

**Stage 7: Data Visualization**

This notebook is independently executable in Google Colab. Every step explains what we do, why we do it, what we obtain, and how the result is used in the next stage.


## Stage Objective

Stage 7 communicates the analysis visually so patterns can be understood faster than from raw tables alone.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path = "/content/Air_Conditioners_Cleaned.csv"
df = pd.read_csv(csv_path)

print("CSV loaded successfully.")
print("Dataset shape:", df.shape)
```

## Why Visualisation

Charts make distributions, category differences and relationships easier to compare and help the audience see the evidence behind the analysis.


```python
nums=df.select_dtypes(include=np.number).columns.tolist(); cats=df.select_dtypes(include=['object','category','bool']).columns.tolist(); print(nums,cats)
```

## Reload the CSV

The CSV is loaded directly so the visualisation stage can be run independently.


```python
if cats:
    df[cats[0]].value_counts().head(15).plot(kind='bar',figsize=(10,5),title=f'Distribution of {cats[0]}'); plt.tight_layout(); plt.show()
```

## Prepare Plot Data

Only the fields required for each chart are selected and aggregated when necessary.


```python
if 'discount_price' in df.columns:
    df['discount_price'].plot(kind='hist',bins=20,figsize=(9,5),title='Price Distribution'); plt.xlabel('discount_price'); plt.tight_layout(); plt.show()
```

## Category Distribution

A bar chart is used for a categorical distribution because bar charts make differences in category counts easy to compare.


```python
if 'ratings' in df.columns:
    df['ratings'].plot(kind='hist',bins=10,figsize=(9,5),title='Rating Distribution'); plt.xlabel('ratings'); plt.tight_layout(); plt.show()
```

## Price Distribution

A histogram is used for **discount_price** to show how observations are distributed across the observed price range.


```python
if cats and nums:
    g=df.groupby(cats[0])[nums[0]].mean().sort_values(ascending=False).head(15); g.plot(kind='bar',figsize=(10,5),title=f'Average {nums[0]} by {cats[0]}'); plt.tight_layout(); plt.show()
```

## Rating Distribution

A histogram is used for **ratings** to show the distribution of ratings.


```python
if nums:
    df[nums[0]].plot(kind='box',figsize=(8,4),title=f'Box Plot of {nums[0]}'); plt.tight_layout(); plt.show()
```

## Group Comparison

A grouped summary is visualised to compare a numerical measure across product categories.


```python
if len(nums)>=2:
    plt.figure(figsize=(8,5)); plt.scatter(df[nums[0]],df[nums[1]],alpha=.5); plt.xlabel(nums[0]); plt.ylabel(nums[1]); plt.title(f'{nums[0]} vs {nums[1]}'); plt.tight_layout(); plt.show()
```

## Relationship Plot

Where two numeric variables are suitable, a scatter plot is used to show their observed relationship.


```python
print('Charts completed with titles, labels and readable layouts.')
```

Each chart is interpreted from the plotted evidence. We describe patterns visible in the data rather than presenting unsupported causal claims.
