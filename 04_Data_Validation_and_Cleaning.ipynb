## Air Conditioners Data Analysis

**Stage 4: Data Validation & Cleaning**

This notebook is independently executable in Google Colab. Every step explains what we do, why we do it, what we obtain, and how the result is used in the next stage.


## Stage Objective

Stage 4 validates the dataset and prepares a reliable analytical DataFrame.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the project CSV directly so this notebook is independently executable.
csv_path1 = "/content/Most_Expensive_Products.csv"
df1 = pd.read_csv(csv_path1)
csv_path2 = "/content/Top_Discounted_Products.csv"
df2 = pd.read_csv(csv_path2)
csv_path3 = "/content/Top_Rated_Products.csv"
df3 = pd.read_csv(csv_path3)

print("CSV loaded successfully.")
print("Dataset shape:", df1.shape)
print("Dataset shape:", df2.shape)
print("Dataset shape:", df3.shape)
```

## Why Validation Is Important

A result can be numerically correct but still misleading if missing, duplicated, malformed or inconsistent values are not checked.


```python
print('Total missing:', int(df1.isna().sum().sum()))
print('Total missing:', int(df2.isna().sum().sum()))
print('Total missing:', int(df3.isna().sum().sum()))
```

## Reload the CSV

The source CSV is loaded directly to keep the stage reproducible.


```python
print('Duplicates:', int(df1.duplicated().sum()))
print('Duplicates:', int(df2.duplicated().sum()))
print('Duplicates:', int(df3.duplicated().sum()))
```

## Missing-Value Check

Every column is checked for missing values. The exact counts are reported rather than assumed.


```python
nums=df1.select_dtypes(include=np.number).columns.tolist(); print(nums)
nums=df2.select_dtypes(include=np.number).columns.tolist(); print(nums)
nums=df3.select_dtypes(include=np.number).columns.tolist(); print(nums)
```

## Duplicate Check

The complete row is checked for duplicates so repeated records can be identified before aggregation.


```python
for c in nums: print(c, 'min=',df1[c].min(),'max=',df1[c].max())
for c in nums: print(c, 'min=',df2[c].min(),'max=',df2[c].max())
for c in nums: print(c, 'min=',df3[c].min(),'max=',df3[c].max())
```

## Numeric Conversion Check

Numeric fields are safely converted where necessary and conversion failures are inspected.


```python
clean_df1=df1.copy()
for c in nums: clean_df1[c]=pd.to_numeric(clean_df1[c],errors='coerce')
print(clean_df1.dtypes)
clean_df2=df2.copy()
for c in nums: clean_df2[c]=pd.to_numeric(clean_df2[c],errors='coerce')
print(clean_df2.dtypes)
clean_df3=df3.copy()
for c in nums: clean_df3[c]=pd.to_numeric(clean_df3[c],errors='coerce')
print(clean_df3.dtypes)
```

## Range Inspection

Minimum and maximum values are checked for numeric fields. This helps identify unexpected ranges that deserve attention.


```python
for c in clean_df1.select_dtypes(include=['object','category']).columns: clean_df1[c]=clean_df1[c].astype('string').str.strip()
print('Text standardization done.')
for c in clean_df2.select_dtypes(include=['object','category']).columns: clean_df2[c]=clean_df2[c].astype('string').str.strip()
print('Text standardization done.')
for c in clean_df3.select_dtypes(include=['object','category']).columns: clean_df3[c]=clean_df3[c].astype('string').str.strip()
print('Text standardization done.')
```

## Text Consistency

Text/category fields are stripped of unnecessary leading and trailing spaces where appropriate so equivalent labels are less likely to split into separate groups.


```python
print(clean_df1.shape)
print(clean_df2.shape)
print(clean_df3.shape)
```

## Clean Working DataFrame

A working DataFrame is created after the validation operations. The original CSV is not overwritten.


```python
print('Remaining missing:', int(clean_df1.isna().sum().sum()))
print('Remaining missing:', int(clean_df2.isna().sum().sum()))
print('Remaining missing:', int(clean_df3.isna().sum().sum()))
```

## Quality Result

The final quality checks are displayed again so the reader can see the state of the data after validation.
